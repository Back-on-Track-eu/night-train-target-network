# 01 — Source extraction (track infrastructure)

Registers every document behind the per-country track infrastructure
parameters and writes `data/sources_register.csv`, which notebooks 02–05
validate every value against. Nothing is calibrated here.

Mirrors `models/compositions/calib/01_source_extraction.ipynb`.

**Source ids are semantic** (`FR-DRR-2027-A52`, not `I23`). A provenance
pointer is read far more often than written, and a semantic id tells a reviewer
what to open. Ids are stable: a superseded document keeps its id and gains a
note rather than being renumbered.

```python
# STDLIB-ONLY cell. See the seed-export contract in calib/README.md.
import csv
from pathlib import Path


def _calib_dir() -> Path:
    """Anchor on the calib folder whatever the kernel's cwd happens to be."""
    cwd = Path.cwd().resolve()
    for cand in (cwd, *cwd.parents):
        if (cand / "resolution.py").exists():
            return cand
    for sub in ("backend/models/infrastructure/calib", "models/infrastructure/calib"):
        if (cwd / sub).is_dir():
            return (cwd / sub).resolve()
    raise RuntimeError("cannot locate models/infrastructure/calib")


CALIB_DIR = _calib_dir()
DATA_DIR = CALIB_DIR / "data"
DATA_DIR.mkdir(exist_ok=True)


def write_data(name: str, fieldnames: list[str], rows: list[dict]) -> None:
    """Write one committed observation table to calib/data/."""
    with open(DATA_DIR / name, "w", newline="", encoding="utf-8") as fh:
        w = csv.DictWriter(fh, fieldnames=fieldnames)
        w.writeheader()
        w.writerows(rows)
    print(f"data/{name}: {len(rows)} rows")


def read_data(name: str) -> list[dict]:
    with open(DATA_DIR / name, newline="", encoding="utf-8") as fh:
        return list(csv.DictReader(fh))


rows: list[tuple] = []
COLUMNS = [
    "source_id",
    "short_id",
    "title",
    "publisher",
    "pub_year",
    "price_basis_year",
    "currency",
    "kind",
    "url_or_file",
    "date_accessed",
    "reliability_note",
]
print(f"calib dir: {CALIB_DIR}")
```
